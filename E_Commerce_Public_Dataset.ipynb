## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Kategori produk apa yang memiliki rata-rata penilaian terendah dan persentase ulasan buruk (1–2 bintang) tertinggi selama September 2016 hingga Oktober 2018, serta kategori mana yang harus diprioritaskan untuk evaluasi dan peningkatan kualitas produk?
  - **Keterangan:**
      1. **Specific:** Berfokus pada evaluasi kualitas produk berdasarkan kategori produk, rata-rata review score, dan persentase ulasan buruk.
      2. **Measurable:** Mengukur rata-rata review score (skala 1–5) dan persentase ulasan 1–2 bintang pada setiap kategori produk.
      3. **Action-Oriented:** Hasil analisis membantu perusahaan menentukan kategori produk yang perlu diprioritaskan untuk evaluasi kualitas, pemeriksaan produk, dan perbaikan pengemasan.
      4. **Relevant:** Kualitas produk yang dipersepsikan pelanggan berhubungan dengan kepuasan pelanggan, reputasi marketplace, dan kemungkinan pembelian kembali
      5. **Time-Bound:** Analisis menggunakan data transaksi dan ulasan dari September 2016 hingga Oktober 2018. Hasilnya digunakan untuk menentukan prioritas perbaikan dalam 6 bulan berikutnya.

- **Pertanyaan 2:** Bagaimana perubahan pendapatan penjualan bulanan (harga produk + biaya pengiriman) pada berbagai kategori produk selama Januari 2017 hingga Agustus 2018, dan berapa tingkat pendapatan yang dapat diperkirakan untuk dua kuartal berikutnya sebagai dasar penentuan alokasi inventaris dan anggaran pemasaran?
  - **Keterangan:**
      1. **Specific:** Berfokus pada perubahan pendapatan penjualan bulanan berdasarkan kategori produk serta perkiraan pendapatan untuk dua kuartal berikutnya.
      2. **Measurable:** Mengukur total pendapatan bulanan, persentase pertumbuhan Month-over-Month (MoM), dan rata-rata pergerakan pendapatan menggunakan 3-month Moving Average.
      3. **Action-Oriented:** Hasil analisis digunakan untuk menentukan kategori produk yang perlu diprioritaskan dalam pengadaan stok serta alokasi anggaran pemasaran
      4. **Relevant:**	Tren pendapatan membantu perusahaan memahami perkembangan penjualan dan menyusun perencanaan bisnis serta pengelolaan inventaris.
      5. **Time-Bound:**	Analisis menggunakan data Januari 2017 hingga Agustus 2018, dengan proyeksi untuk dua kuartal berikutnya setelah periode pengamatan.

- **Pertanyaan 3:** Kelompok pelanggan apa saja yang dapat diidentifikasi berdasarkan Recency dan Monetary Value selama September 2016 hingga September 2018, berapa kontribusi pendapatan dari setiap kelompok, dan kelompok pelanggan mana yang harus diprioritaskan untuk kampanye mempertahankan serta mengaktifkan kembali pelanggan dalam 6 bulan berikutnya?
  - **Keterangan:**
      1. **Specific:**	Berfokus pada pengelompokan pelanggan berdasarkan Recency (waktu sejak transaksi terakhir) dan Monetary Value (total nilai pembelian).
      2. **Measurable:**	Mengukur jumlah pelanggan, rata-rata nilai pembelian, persentase kontribusi pendapatan, dan jumlah pelanggan dalam setiap segmen.
      3. **Action-Oriented:**	Hasil analisis membantu perusahaan menentukan strategi pemasaran berdasarkan segmen, seperti memberikan promosi kepada pelanggan potensial dan menjalankan kampanye win-back kepada pelanggan yang sudah lama tidak bertransaksi.
      4. **Relevant:**	Segmentasi pelanggan membantu perusahaan memahami karakteristik dan perilaku pembelian pelanggan sehingga strategi pemasaran dapat dilakukan secara lebih efektif.
      5. **Time-Bound:**	Menggunakan data transaksi September 2016 hingga September 2018, dengan target implementasi strategi pemasaran selama 6 bulan berikutnya.

B. REPEAT CUSTOMERS

```python
# MERGING ORDERS WITH CUSTOMERS TO GET THE REAL CUSTOMER IDENTITY (CUSTOMER_UNIQUE_ID)
orders_customers_df = orders_df.merge(customers_df, on='customer_id', how='inner')

# REMOVING ORDERS THAT NEVER GENERATED A TRANSACTION
orders_customers_df = orders_customers_df[~orders_customers_df['order_status'].isin(['canceled', 'unavailable'])]

# COUNTING HOW MANY ORDERS EACH REAL CUSTOMER HAS PLACED
orders_per_customer = orders_customers_df.groupby('customer_unique_id')['order_id'].nunique()

# CALCULATING THE OVERALL REPEAT CUSTOMER RATE
print("=== REPEAT CUSTOMER RATE (%) ===")
print((orders_per_customer > 1).mean() * 100)
```

```python
# SELECTING THE MAIN PAYMENT TYPE OF EACH ORDER (THE ONE WITH THE HIGHEST PAYMENT VALUE)
payment_main = (order_payments_df.sort_values('payment_value', ascending=False)
                .drop_duplicates('order_id')[['order_id', 'payment_type']])

# SELECTING THE FIRST PRODUCT CATEGORY OF EACH ORDER
first_item_df = (order_items_df.sort_values(['order_id', 'order_item_id'])
                 .drop_duplicates('order_id')[['order_id', 'product_id']]
                 .merge(products_with_translation_df[['product_id', 'product_category_name_english']],
                        on='product_id', how='left'))
first_item_df['product_category_name_english'] = first_item_df['product_category_name_english'].fillna('Unknown')

# KEEPING ONLY THE FIRST ORDER OF EACH CUSTOMER AND ATTACHING PAYMENT TYPE AND CATEGORY
first_order_df = (orders_customers_df.sort_values('order_purchase_timestamp')
                  .drop_duplicates('customer_unique_id')
                  .merge(payment_main, on='order_id', how='left')
                  .merge(first_item_df[['order_id', 'product_category_name_english']], on='order_id', how='left'))

# LABELING EACH CUSTOMER AS REPEAT (MORE THAN ONE ORDER) OR NOT
first_order_df['is_repeat'] = first_order_df['customer_unique_id'].map(orders_per_customer > 1)

# CREATING A FUNCTION TO CALCULATE THE REPEAT RATE OF EACH GROUP (MINIMUM 100 CUSTOMERS PER GROUP)
def calculate_repeat_rate(df, group_col, min_customers=100):
    # CALCULATING THE REPEAT RATE AND THE NUMBER OF CUSTOMERS IN EACH GROUP
    summary = (df.groupby(group_col)['is_repeat']
               .agg(repeat_rate='mean', customers='count')
               .reset_index())

    # REMOVING SMALL GROUPS SO THEY DO NOT APPEAR AS TOP PERFORMERS BY CHANCE
    summary = summary[summary['customers'] >= min_customers].copy()

    # CONVERTING THE REPEAT RATE INTO PERCENTAGE AND SORTING FROM HIGHEST TO LOWEST
    summary['repeat_rate'] = summary['repeat_rate'] * 100
    return summary.sort_values('repeat_rate', ascending=False).reset_index(drop=True)

# CREATING THE REPEAT RATE BY CUSTOMER STATE
repeat_by_state = calculate_repeat_rate(first_order_df, 'customer_state')

# CREATING THE REPEAT RATE BY PAYMENT TYPE
repeat_by_payment = calculate_repeat_rate(first_order_df, 'payment_type')

# CREATING THE REPEAT RATE BY FIRST-PURCHASE CATEGORY
repeat_by_category = calculate_repeat_rate(first_order_df, 'product_category_name_english')

# SAVING THE OVERALL REPEAT RATE (%) TO BE USED AS A REFERENCE LINE IN THE CHARTS
overall_repeat_rate = (orders_per_customer > 1).mean() * 100

# DISPLAYING THE RESULTS TO VERIFY THEM
print("=== OVERALL REPEAT RATE (%) ===")
print(overall_repeat_rate)

print("\n=== REPEAT RATE (%) BY CUSTOMER STATE ===")
display(repeat_by_state.head(10))

print("\n=== REPEAT RATE (%) BY PAYMENT TYPE ===")
display(repeat_by_payment)

print("\n=== REPEAT RATE (%) BY FIRST-PURCHASE CATEGORY ===")
display(repeat_by_category.head(10))
```

```python
# MONTHLY REVENUE (PRICE + FREIGHT) JANUARY 2017 - AUGUST 2018, CANCELED/UNAVAILABLE ORDERS ALREADY REMOVED IN REVENUE_DF
# SEPTEMBER-OCTOBER 2018 ARE EXCLUDED BECAUSE THEY CONTAIN ONLY 20 ORDERS (INCOMPLETE DATA, SEE THE MONTHLY ORDER VOLUME ABOVE)
rev_period_df = revenue_df[(revenue_df['order_purchase_timestamp'] >= '2017-01-01') &
                           (revenue_df['order_purchase_timestamp'] < '2018-09-01')].copy()
rev_period_df['year_month'] = rev_period_df['order_purchase_timestamp'].dt.to_period('M')

# TOTAL REVENUE PER MONTH, MONTH-OVER-MONTH (MoM) GROWTH, AND 3-MONTH MOVING AVERAGE
monthly_total = rev_period_df.groupby('year_month')['revenue'].sum().to_frame('revenue')
monthly_total['mom_growth_pct'] = monthly_total['revenue'].diff() / monthly_total['revenue'].shift(1) * 100
monthly_total['ma3'] = monthly_total['revenue'].rolling(3).mean()
display(monthly_total.round(1))

# THE TOP 5 CATEGORIES BY REVENUE IN THE WHOLE PERIOD
top5_categories = (rev_period_df.groupby('product_category_name_english')['revenue']
                   .sum().nlargest(5).index.tolist())
monthly_top5 = (rev_period_df[rev_period_df['product_category_name_english'].isin(top5_categories)]
                .pivot_table(index='year_month', columns='product_category_name_english', values='revenue', aggfunc='sum')[top5_categories])
monthly_top5_ma3 = monthly_top5.rolling(3).mean()
monthly_top5_mom = monthly_top5.diff() / monthly_top5.shift(1) * 100

print('\n=== TOP 5 CATEGORIES: SHARE OF TOTAL REVENUE ===')
top5_share = (monthly_top5.sum() / monthly_total['revenue'].sum() * 100).round(1)
display(pd.DataFrame({'revenue': monthly_top5.sum().round(0), 'share_pct': top5_share}))

print('\n=== TOP 5 CATEGORIES: AVERAGE MoM GROWTH (%) IN 2018 VS 2017 ===')
display(pd.DataFrame({'2017': monthly_top5_mom.loc['2017-02':'2017-12'].mean(),
                      '2018': monthly_top5_mom.loc['2018-01':'2018-08'].mean()}).round(1))
```

```python
# CHOOSING A FORECAST METHOD: BACKTEST ON THE LAST 3 AND 6 KNOWN MONTHS (TRAIN ON THE EARLIER MONTHS, COMPARE WITH THE REAL VALUES)
y = monthly_total['revenue'].values.astype(float)

def mape(actual, predicted):
    return np.mean(np.abs(actual - predicted) / actual) * 100

def forecast_methods(train, horizon):
    x_train = np.arange(len(train))
    x_future = np.arange(len(train), len(train) + horizon)
    linear_all = np.polyval(np.polyfit(x_train, train, 1), x_future)
    last6 = train[-6:]
    linear_last6 = np.polyval(np.polyfit(np.arange(6), last6, 1), np.arange(6, 6 + horizon))
    flat_ma3 = np.repeat(train[-3:].mean(), horizon)
    return {'Linear trend (all months)': linear_all,
            'Linear trend (last 6 months)': linear_last6,
            '3-month moving average (flat)': flat_ma3}

backtest_rows = []
for horizon in (3, 6):
    train, actual = y[:-horizon], y[-horizon:]
    for name, pred in forecast_methods(train, horizon).items():
        backtest_rows.append({'holdout_months': horizon, 'method': name, 'MAPE_%': round(mape(actual, pred), 1)})
backtest_df = pd.DataFrame(backtest_rows).pivot(index='method', columns='holdout_months', values='MAPE_%')
print('=== BACKTEST ERROR (MAPE %, LOWER IS BETTER) ===')
display(backtest_df)

# FORECASTING SEPTEMBER 2018 - FEBRUARY 2019 (2 NEXT QUARTERS) WITH THE BEST METHOD: THE 3-MONTH MOVING AVERAGE
CHOSEN_METHOD = '3-month moving average (flat)'
error_band = backtest_df.loc[CHOSEN_METHOD].mean() / 100
future_months = pd.period_range('2018-09', periods=6, freq='M')
forecast_total = pd.DataFrame({'forecast': forecast_methods(y, 6)[CHOSEN_METHOD]}, index=future_months)
forecast_total['lower'] = forecast_total['forecast'] * (1 - error_band)
forecast_total['upper'] = forecast_total['forecast'] * (1 + error_band)

forecast_quarters = pd.DataFrame({
    'quarter': ['Sep-Nov 2018', 'Dec 2018-Feb 2019'],
    'forecast': [forecast_total['forecast'].iloc[:3].sum(), forecast_total['forecast'].iloc[3:].sum()],
    'lower': [forecast_total['lower'].iloc[:3].sum(), forecast_total['lower'].iloc[3:].sum()],
    'upper': [forecast_total['upper'].iloc[:3].sum(), forecast_total['upper'].iloc[3:].sum()]}).round(0)
print('\nError band used (average backtest MAPE): +/-', round(error_band * 100, 1), '%')
print('\n=== FORECAST OF TOTAL REVENUE FOR THE NEXT 2 QUARTERS ===')
display(forecast_quarters)

# THE SAME METHOD FOR EACH OF THE TOP 5 CATEGORIES (NEXT QUARTER = 3 x THE LAST 3-MONTH AVERAGE)
forecast_top5 = pd.DataFrame({
    'last_3_month_avg': monthly_top5.tail(3).mean(),
    'forecast_per_quarter': monthly_top5.tail(3).mean() * 3})
forecast_top5['share_of_total_forecast_pct'] = forecast_top5['forecast_per_quarter'] / forecast_quarters['forecast'].iloc[0] * 100
forecast_top5 = forecast_top5.round(1)
print('\n=== FORECAST PER QUARTER FOR THE TOP 5 CATEGORIES ===')
display(forecast_top5)
```

#### Pertanyaan 3: segmentasi pelanggan berdasarkan Recency dan Monetary (RFM)

```python
# BUILDING THE RFM TABLE (RECENCY + MONETARY) FOR SEPTEMBER 2016 - SEPTEMBER 2018
SNAPSHOT_DATE = pd.Timestamp('2018-10-01')   # THE DAY AFTER THE END OF THE ANALYSIS PERIOD

# 1. ORDERS THAT GENERATED REVENUE, LINKED TO THE REAL CUSTOMER IDENTITY (CUSTOMER_UNIQUE_ID)
rfm_orders_df = (orders_df.merge(customers_df[['customer_id', 'customer_unique_id']], on='customer_id', how='inner'))
rfm_orders_df = rfm_orders_df[(~rfm_orders_df['order_status'].isin(['canceled', 'unavailable'])) &
                              (rfm_orders_df['order_purchase_timestamp'] >= '2016-09-01') &
                              (rfm_orders_df['order_purchase_timestamp'] < SNAPSHOT_DATE)]

# 2. VALUE OF EACH ORDER = SUM OF (PRICE + FREIGHT) OF ITS ITEMS
order_value_df = (order_items_df.assign(order_value=order_items_df['price'] + order_items_df['freight_value'])
                  .groupby('order_id', as_index=False)['order_value'].sum())
rfm_orders_df = rfm_orders_df.merge(order_value_df, on='order_id', how='inner')

# 3. ONE ROW PER CUSTOMER: LAST PURCHASE, NUMBER OF ORDERS, AND TOTAL SPENDING
rfm_df = (rfm_orders_df.groupby('customer_unique_id')
          .agg(last_purchase=('order_purchase_timestamp', 'max'),
               frequency=('order_id', 'nunique'),
               monetary=('order_value', 'sum'))
          .reset_index())
rfm_df['recency'] = (SNAPSHOT_DATE - rfm_df['last_purchase']).dt.days

print('Customers analysed:', len(rfm_df))
print('\n=== SHARE OF CUSTOMERS BY NUMBER OF ORDERS (WHY FREQUENCY IS NOT USED) ===')
display((rfm_df['frequency'].value_counts(normalize=True).head(4) * 100).round(2))
display(rfm_df[['recency', 'monetary']].describe().round(1))
```

```python
# SCORING RECENCY AND MONETARY IN QUARTILES (4 = BEST). RANK(METHOD='FIRST') AVOIDS DUPLICATE BIN EDGES
rfm_df['R_score'] = pd.qcut(rfm_df['recency'].rank(method='first'), 4, labels=[4, 3, 2, 1]).astype(int)
rfm_df['M_score'] = pd.qcut(rfm_df['monetary'].rank(method='first'), 4, labels=[1, 2, 3, 4]).astype(int)

print('=== RANGE OF EACH SCORE ===')
display(rfm_df.groupby('R_score')['recency'].agg(['min', 'max']).rename(columns={'min': 'recency_min_days', 'max': 'recency_max_days'}))
display(rfm_df.groupby('M_score')['monetary'].agg(['min', 'max']).round(1))

# SEGMENT RULES (RECENCY SCORE 3-4 = RECENT, MONETARY SCORE 3-4 = HIGH VALUE)
def assign_segment(row):
    if row['R_score'] >= 3 and row['M_score'] >= 3:
        return 'Champions'
    if row['R_score'] >= 3:
        return 'Potential Customers'
    if row['M_score'] >= 3:
        return 'At-Risk Big Spenders'
    return 'Lost Low-Value'

rfm_df['segment'] = rfm_df.apply(assign_segment, axis=1)

segment_order = ['Champions', 'At-Risk Big Spenders', 'Potential Customers', 'Lost Low-Value']
segment_summary = (rfm_df.groupby('segment')
                   .agg(customers=('customer_unique_id', 'count'),
                        avg_recency_days=('recency', 'mean'),
                        avg_monetary=('monetary', 'mean'),
                        total_revenue=('monetary', 'sum'))
                   .reindex(segment_order))
segment_summary['pct_customers'] = segment_summary['customers'] / segment_summary['customers'].sum() * 100
segment_summary['pct_revenue'] = segment_summary['total_revenue'] / segment_summary['total_revenue'].sum() * 100
print('\n=== RFM SEGMENT SUMMARY ===')
display(segment_summary.round(1))
```

## Analisis Lanjutan (Opsional)

```python
# ADVANCED ANALYSIS: DOES LATE DELIVERY EXPLAIN THE BAD REVIEWS OF THE PRIORITY CATEGORIES?
# LINKING THE REVIEW DATA (QUESTION 1) WITH THE DELIVERY RESULT OF EACH ORDER (DELIVERED_ORDERS FROM THE EDA)
delay_review_df = category_review_df.merge(delivered_orders[['order_id', 'is_late', 'delay_days']], on='order_id', how='inner')
delay_review_df['delay_bucket'] = pd.cut(delay_review_df['delay_days'], bins=[-np.inf, 0, 3, 7, np.inf],
                                         labels=['On time', '1-3 days late', '4-7 days late', '>7 days late'])

# 1. BAD REVIEW % BY HOW LATE THE ORDER WAS (ONE ROW PER ORDER)
bucket_df = (delay_review_df.drop_duplicates('order_id')
             .groupby('delay_bucket', observed=True)
             .agg(orders=('order_id', 'count'), bad_review_pct=('is_bad_review', 'mean'), avg_score=('review_score', 'mean')))
bucket_df['bad_review_pct'] = bucket_df['bad_review_pct'] * 100

# 2. FOR THE TOP 5 PRIORITY CATEGORIES: BAD REVIEW % WHEN ON TIME VS WHEN LATE
top5_priority = category_quality.head(5)['product_category_name_english'].tolist()
cat_delay_df = (delay_review_df[delay_review_df['product_category_name_english'].isin(top5_priority)]
                .groupby(['product_category_name_english', 'is_late'])['is_bad_review'].agg(['mean', 'count']).reset_index())
cat_delay_df['mean'] = cat_delay_df['mean'] * 100
cat_delay_pivot = cat_delay_df.pivot(index='product_category_name_english', columns='is_late', values='mean').reindex(top5_priority)
cat_delay_pivot.columns = ['bad_review_pct_on_time', 'bad_review_pct_late']

# 3. SHARE OF EACH CATEGORY'S ORDERS THAT ARRIVED LATE, AND THE SHARE OF ITS BAD REVIEWS THAT CAME FROM LATE ORDERS
late_share = delay_review_df[delay_review_df['product_category_name_english'].isin(top5_priority)].groupby('product_category_name_english').apply(
    lambda d: pd.Series({'late_orders_pct': d['is_late'].mean() * 100,
                         'bad_reviews_from_late_orders_pct': d.loc[d['is_bad_review'], 'is_late'].mean() * 100}),
    include_groups=False).reindex(top5_priority)
cat_delay_pivot = cat_delay_pivot.join(late_share)

print('=== BAD REVIEW % BY DELIVERY DELAY ===')
display(bucket_df.round(1))
print('\n=== TOP 5 PRIORITY CATEGORIES: IS LATE DELIVERY THE CAUSE? ===')
display(cat_delay_pivot.round(1))

fig, axes = plt.subplots(1, 2, figsize=(16, 5.5))
axes[0].bar(bucket_df.index.astype(str), bucket_df['bad_review_pct'], color=['#27ae60', '#f1c40f', '#e67e22', '#c0392b'])
axes[0].bar_label(axes[0].containers[0], fmt='%.1f%%', padding=3)
axes[0].set_title('Bad review % by delivery delay'); axes[0].set_ylabel('% bad reviews (1-2 stars)')
axes[0].set_ylim(0, bucket_df['bad_review_pct'].max() * 1.15)

x = np.arange(len(cat_delay_pivot)); w = 0.38
axes[1].bar(x - w/2, cat_delay_pivot['bad_review_pct_on_time'], w, label='Delivered on time', color='#27ae60')
axes[1].bar(x + w/2, cat_delay_pivot['bad_review_pct_late'], w, label='Delivered late', color='#c0392b')
axes[1].set_xticks(x); axes[1].set_xticklabels(cat_delay_pivot.index, rotation=20, ha='right')
axes[1].set_title('Top 5 priority categories: bad review % on time vs late'); axes[1].legend()
plt.tight_layout(); plt.show()
```
